<span style="font-size:20px">**Perform data cleaning (Replace ? with NaN)**</span>

```python
data.tail(20)
```

```python
import numpy as np
```

```python
data.columns
```

```python
#Go to the workclass column and replace '?' with NaN
#Show updated values temporarily
#But original data stays unchanged
#Now store the updated column back into the dataset
# we are replacing ? to nan because pandas doesnot treat ? as null value
data['workclass']=data['workclass'].replace('?',np.nan)
data['occupation']=data['occupation'].replace('?',np.nan)
data['native-country']=data['native-country'].replace('?',np.nan)
```

<span style="font-size:20px">**Drop all the missing values**</span>

```python
#It calculates the percentage of missing values in each column
#len(data)-->Total number of rows in the dataset
#For each column, what % of values are missing?
per_missing=data.isnull().sum()*100/len(data)
```

```python
#dropna()-->Function used to remove missing values
#how='any'-->if a row has even ONE missing value, delete the entire row
#inplace=True-->Makes changes directly in the original dataset

data.dropna(how='any',inplace=True)
data
```

<span style="font-size:20px">**Replace income values[<=50,>50k] with 0 and 1**</span>

```python
data.columns
```

```python
data['income'].unique()
```

```python
data['income'].value_counts()
```

```python
def income_val(val):
    if val=='<=50K':
        return 0
    else:
        return 1
```

```python
data['encoded_income']=data['income'].apply(income_val)
```

```python
data.head(1)
```

```python
#another method
data.replace(to_replace=['<=50K', '>50K'],value=[0,1],inplace=True)
```

<span style="font-size:20px">**Which workclass getting the highest salary**</span>

```python
#It calculates the average income for each workclass and sorts it from highest to lowest
#Groups the dataset based on workclass#Selects only the income column from each group
#Calculates the average income for each workclass
#Highest average income → Lowest
data.groupby('workclass')['income'].mean().sort_values(ascending=False)
```
